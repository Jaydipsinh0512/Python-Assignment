```python
# 1

import requests

url = "https://www.flipkart.com/account/orders"

session = requests.Session()

response1 = session.get(url, timeout=15)
print("First Status Code:", response1.status_code)

response2 = session.get(url, timeout=15)
print("Second Status Code:", response2.status_code)

print("Session Cookies:", session.cookies.get_dict())
```

```python
# 2

import requests

url = "https://api.openweathermap.org/data/2.5/weather"

params = {
    "q": "Ahmedabad",
    "appid": "YOUR_API_KEY",
    "units": "metric"
}

response = requests.get(url, params=params)
data = response.json()

if response.status_code == 200:
    temperature = data["main"]["temp"]
    print("Current Temperature:", temperature, "°C")
else:
    print("Error:", data.get("message", "Request failed"))
```

```python
# 3

import asyncio
import httpx

async def fetch_data(url):
    async with httpx.AsyncClient() as client:
        response = await client.get(url, timeout=15)
        response.raise_for_status()
        return response.json()

async def main():
    posts, users = await asyncio.gather(
        fetch_data("https://jsonplaceholder.typicode.com/posts"),
        fetch_data("https://jsonplaceholder.typicode.com/users")
    )

    print("First 5 Posts:")
    for post in posts[:5]:
        print(post["title"])

    print("\nFirst 5 Usernames:")
    for user in users[:5]:
        print(user["username"])

await main()
```

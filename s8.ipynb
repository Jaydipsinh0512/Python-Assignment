```python
# 2

steps = [7500, 8200, 9100, 10500, 12000, 6000, 9800]

i = 0
while i < len(steps):
    if steps[i] > 10000:
        print("First day above 10,000 steps: Day", i + 1)
        break
    i += 1
```

```python
# 3

def print_long_teams(teams):
    for team in teams:
        if len(team) <= 6:
            continue
        print(team)

ipl_teams = ["MI", "Chennai Super Kings", "Rajasthan Royals", "RCB", "Gujarat Titans"]
print_long_teams(ipl_teams)
```

```python
# 4

durations = [210, 185, 240, 195, 260]

for position, duration in enumerate(durations, start=1):
    print(f"Song {position}: {duration} seconds")
```

```python
# 5

prices = [499, 0, 799, 650, 0, 399, 899]
total = 0

for price in prices:
    if price == 0:
        continue
    total += price
    if total > 2000:
        break

print("Final total: ₹", total)
```

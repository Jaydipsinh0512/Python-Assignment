```python
# 1

class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        return self.price - (self.price * 10 / 100)

shirt = Product("Shirt", 1100)
print(shirt.name, "-", shirt.get_discounted_price())
```

```python
# 2

class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        return self.price - (self.price * 10 / 100)

class Electronics(Product):
    def get_discounted_price(self):
        return self.price - (self.price * 20 / 100)

phone = Electronics("Smartphone", 25000)
print(phone.name, "-", phone.get_discounted_price())
```

```python
# 3

def show_final_price(item):
    print(f"{item.name}: ₹{item.get_discounted_price()}")

shirt = Product("Shirt", 1100)
phone = Electronics("Smartphone", 25000)

show_final_price(shirt)
show_final_price(phone)
```

```python
# 4

class Ticket:
    def __init__(self, base_price):
        self.base_price = base_price

    def get_final_price(self):
        return self.base_price

class PremiumTicket(Ticket):
    def get_final_price(self):
        return super().get_final_price() + 50

normal_ticket = Ticket(250)
premium_ticket = PremiumTicket(250)

print("Normal ticket price: ₹", normal_ticket.get_final_price())
print("Premium ticket price: ₹", premium_ticket.get_final_price())
```

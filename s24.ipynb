```python
# 1

import requests

url = "https://jsonplaceholder.typicode.com/posts"

data = {
    "title": "My First Post",
    "body": "Hello from Python!",
    "userId": 1
}

response = requests.post(url, json=data)

print("Status Code:", response.status_code)
print("JSON Data:", response.json())
```

```python
# 2

import requests

url = "https://jsonplaceholder.typicode.com/posts"

name = input("Enter playlist name: ")
description = input("Enter playlist description: ")

data = {
    "title": name,
    "body": description,
    "userId": 1
}

response = requests.post(url, json=data)
result = response.json()

print("Playlist ID:", result["id"])
```

```python
# 3
import requests

url = "https://jsonplaceholder.typicode.com/posts"

data = {
    "title": "Jaydip",
    "body": "Data Scientist",
    "userId": 1
}

try:
    response = requests.post(url, json=data, timeout=15)

    print("Status Code:", response.status_code)

    if response.status_code == 201:
        result = response.json()
        print("User ID:", result.get("id"))
        print("Created At:", result.get("createdAt", "Not provided"))
    else:
        print("Request failed:", response.text)

except requests.exceptions.RequestException as e:
    print("Connection Error:", e)
```

```python
# 4

import requests
import csv

url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)
posts = response.json()[:5]

with open("posts.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["title", "userId"])

    for post in posts:
        writer.writerow([post["title"], post["userId"]])

print("Data saved to posts.csv")
```

```python
# 5

import requests
import json

url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)
posts = response.json()[:5]

data = []

for post in posts:
    data.append({
        "title": post["title"],
        "userId": post["userId"]
    })

with open("posts.json", "w", encoding="utf-8") as file:
    json.dump(data, file, indent=4)

print("Data saved to posts.json")
```

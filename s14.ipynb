```python
# 2

import random

def playlist_generator(songs):
    shuffled = songs[:]
    random.shuffle(shuffled)
    for song in shuffled:
        yield song

playlist = ["Kesariya", "Apna Bana Le", "Tum Hi Ho", "Raataan Lambiyan"]

for song in playlist_generator(playlist):
    print(song)
```

```python
# 3

cart = ['Pizza', 'Burger', 'Fries', 'Coke']

for index, item in enumerate(cart, start=1):
    print(index, item)
```

```python
# 5

def order_id_generator():
    order_id = 1001
    while True:
        yield order_id
        order_id += 1

order_ids = order_id_generator()

print(next(order_ids))
print(next(order_ids))
print(next(order_ids))
```

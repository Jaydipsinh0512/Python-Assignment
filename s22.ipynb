```python
# 1

import requests

url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)
posts = response.json()

for post in posts[:5]:
    print(post["title"])
```

```python
# 2

import json

restaurant = {
    "name": "Spicy Hub",
    "location": "Ahmedabad",
    "cuisines": ["Gujarati", "North Indian"],
    "ratings": 4.5
}

json_data = json.dumps(restaurant, indent=4)

print(json_data)
```

```python
# 3

import requests

url = "https://jsonplaceholder.typicode.com/posts"

data = {
    "title": "My Favorite Playlist",
    "userId": 1,
    "body": "My favorite songs collection"
}

response = requests.post(url, json=data)

print("Status Code:", response.status_code)
print("Response:", response.json())
```

```python
# 4

import requests

url = "https://jsonplaceholder.typicode.com/posts"

params = {"userId": 2}

response = requests.get(url, params=params)
posts = response.json()

for post in posts:
    print(post["id"])
```

```python
# 5

import requests

url = "https://jsonplaceholder.typicode.com/posts"

headers = {
    "Authorization": "Bearer YOUR_ACCESS_TOKEN",
    "Accept": "application/json"
}

response = requests.get(url, headers=headers)

print("Status Code:", response.status_code)
```
